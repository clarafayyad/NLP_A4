```python
import stanza
stanza.download('pl')
nlp = stanza.Pipeline('pl')
```

```python
doc = nlp("Jagoda lubi kawę, a Julianna herbatę.")
for sentence in doc.sentences:
    print("=== Dependency Tree ===")
    for word in sentence.words:
        print(f"word: {word.text}\thead: {word.head}\tdeprel: {word.deprel}")
```
